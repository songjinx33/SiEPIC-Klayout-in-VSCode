```python
# ============================================================
# Global/default GC settings
# Define once and reuse in later cells
# ============================================================
GC_SETTINGS = {
    "gc_cell_name": "SiEPICfab_1550TE",
    "gc_library": gclib_name,
    "gc_pitch": 127.0,   # microns
    "default_num_gc": 4,
    "label_location": 2,
}
WG_SETTINGS = {
    "wg_type": "Strip TE 1550 nm, w=500 nm (core-clad)",
}
device_prefix = "opt_in_TE_1550_device_JSong"
```

```python
def gc_loopback(
    x_pos_GC=0,
    y_pos_GC=None,
    cellname="gc_loopback",
    device_label="opt_in_TE_1550_device_JSong_Loopback",
    xmax=None,
    num_gc=None,
    port_pairs= [(0, 1)],
    x_spacing=20.0,
    row_spacing=None,
    turtle_A=None,
):
    """
    Place a GC array and connect selected GC pairs with loopback waveguides.

    Parameters
    ----------
    x_pos_GC, y_pos_GC : float
        GC array anchor position in microns.

    cellname : str
        New cell name.

    device_label : str
        Label attached to the GC array.

    xmax : float
        Maximum x position before wrapping to next row.

    num_gc : int
        Number of GCs in the array.

    port_pairs : list of tuple[int, int]
        GC index pairs to connect.
        Examples:
            [(0, 1)]             -> connect GC0 and GC1
            [(0, 1), (2, 3)]     -> connect GC0-GC1 and GC2-GC3

    turtle_A : list or list of lists or None
        Routing path(s) for connect_pins_with_waveguide.
        - If a single list is given, use the same turtle for all pairs.
        - If a list of lists is given, each pair uses its own turtle.

    Returns
    -------
    next_x, next_y, xmax
    """

    global GC_SETTINGS, WG_SETTINGS

    gc_cell_name   = GC_SETTINGS["gc_cell_name"]
    gc_library     = GC_SETTINGS["gc_library"]
    gc_pitch       = GC_SETTINGS["gc_pitch"]
    label_location = GC_SETTINGS["label_location"]

    if num_gc is None:
        num_gc = GC_SETTINGS["default_num_gc"]

    if y_pos_GC is None:
        y_pos_GC = top_cell.bbox().height() * dbu

    if xmax is None:
        xmax = top_cell.bbox().width() * dbu

    if row_spacing is None:
        row_spacing = (num_gc - 1) * gc_pitch + 2 * wg_bend_radius

    # if port_pairs is None:
    #     port_pairs = [(0, 1)]

    if turtle_A is None:
        turtle_A = []

    wg_type = WG_SETTINGS["wg_type"]
    gc_ports = ["opt1"] * num_gc

    if num_gc < 2:
        raise ValueError("num_gc must be at least 2")

    # validate port_pairs
    for pair in port_pairs:
        if len(pair) != 2:
            raise ValueError(f"Each entry in port_pairs must have 2 indices, got {pair}")
        i, j = pair
        if not (0 <= i < num_gc and 0 <= j < num_gc):
            raise ValueError(f"Invalid GC pair {pair} for num_gc={num_gc}")

    # normalize turtle_A
    # case 1: single turtle for all pairs, e.g. [50, -90, 127, 90, 50]
    # case 2: one turtle per pair, e.g. [[...], [...]]
    if len(port_pairs) == 1:
        turtle_list = [turtle_A]
    else:
        if isinstance(turtle_A, list) and len(turtle_A) > 0 and isinstance(turtle_A[0], list):
            if len(turtle_A) != len(port_pairs):
                raise ValueError("Number of turtle paths must match number of port pairs")
            turtle_list = turtle_A
        else:
            turtle_list = [turtle_A for _ in port_pairs]

    # -----------------------------
    # Create new cell
    # -----------------------------
    cell = ly.create_cell(cellname)
    top_cell.insert(CellInstArray(cell.cell_index(), Trans(Trans.R0, 0, 0)))

    # -----------------------------
    # Load GC cell
    # -----------------------------
    cell_gc = create_cell2(ly, gc_cell_name, gc_library)
    if cell_gc is None:
        raise ValueError(f"Could not load GC cell '{gc_cell_name}' from library '{gc_library}'")

    # -----------------------------
    # Auto place first device near top-left
    # -----------------------------
    if x_pos_GC == 0 and y_pos_GC == top_cell.bbox().height() * dbu:
        x_pos_GC = cell_gc.bbox().width() * dbu
        y_pos_GC -= (num_gc - 1) * gc_pitch + cell_gc.bbox().height() * dbu / 2

    # -----------------------------
    # Place GC array
    # -----------------------------
    inst_GCs = coupler_array(
        cell,
        cell_name=gc_cell_name,
        cell_library=gc_library,
        x_offset=x_pos_GC / dbu,
        y_offset=y_pos_GC / dbu,
        label=device_label,
        label_location=label_location,
        count=num_gc,
    )

    # -----------------------------
    # Connect selected GC pairs
    # -----------------------------
    for (i, j), turtle in zip(port_pairs, turtle_list):
        connect_pins_with_waveguide(
            inst_GCs[i], gc_ports[i],
            inst_GCs[j], gc_ports[j],
            # turtle_A=turtle,
            waveguide_type=wg_type,
            r=wg_bend_radius
        )

    # -----------------------------
    # Suggested next placement
    # -----------------------------
    next_x = cell.bbox().right * dbu + x_spacing + cell_gc.bbox().width() * dbu
    next_y = y_pos_GC

    if next_x + cell.bbox().width() * dbu > xmax:
        next_x = cell_gc.bbox().width() * dbu
        next_y = cell.bbox().bottom * dbu - row_spacing - cell_gc.bbox().height() * dbu / 2

    return next_x, next_y, xmax
```

```python

def ArbSplitter_cascaded(
    x_pos_GC=0,
    y_pos_GC=None,
    cellname="cellname",
    device_label="opt_in_TE_1550_device_JSong_OneS",
    xmax=None,
    num_gc=2,
    num_dev = 4,
    device_cell_name="y_d8_split9901",
    device_library=lib_name,
    device_params=None,
    device_rotation=Trans.R0,
    device_x=None,
    device_y=None,
    gc_ports=None,
    wg_type = None,
    x_spacing=20.0,
    row_spacing=None,
    routing_to_device=None,
    routing_from_device=None,
):
    """
    Generic placement function for:
        GC array -> routing -> device -> routing

    Notes
    -----
    1. GC information is taken from the global GC_SETTINGS dictionary.
    2. This function assumes the device has optical ports such as opt1, opt2.
    3. The first and last GC are used for input/output by default.
    """

    global GC_SETTINGS, WG_SETTINGS

    if device_params is None:
        device_params = {}

    # if gc_ports is None:
    #     gc_ports = ["opt1", "opt1"]

    if routing_to_device is None:
        routing_to_device = []

    if routing_from_device is None:
        routing_from_device = []

    # Use global GC settings
    gc_cell_name = GC_SETTINGS["gc_cell_name"]
    gc_library = GC_SETTINGS["gc_library"]
    gc_pitch = GC_SETTINGS["gc_pitch"]
    label_location = GC_SETTINGS["label_location"]


    if num_gc is None:
        num_gc = GC_SETTINGS["default_num_gc"]
    gc_ports = ["opt1"] * num_gc
    
    if wg_type is None:
        wg_type = WG_SETTINGS["wg_type"]

    if y_pos_GC is None:
        y_pos_GC = top_cell.bbox().height() * dbu

    if xmax is None:
        xmax = top_cell.bbox().width() * dbu

    if row_spacing is None:
        row_spacing = (num_gc - 1) * gc_pitch + 2 * wg_bend_radius

    # --------------------------------------------------------
    # Create new cell
    # --------------------------------------------------------
    cell = ly.create_cell(cellname)
    top_cell.insert(CellInstArray(cell.cell_index(), Trans(Trans.R0, 0, 0)))

    # --------------------------------------------------------
    # Load GC cell
    # --------------------------------------------------------
    cell_gc = create_cell2(ly, gc_cell_name, gc_library)

    # --------------------------------------------------------
    # Auto place first device near top-left
    # --------------------------------------------------------
    if x_pos_GC == 0 and y_pos_GC == top_cell.bbox().height() * dbu:
        x_pos_GC = int(cell_gc.bbox().width() * dbu)
        y_pos_GC -= int((num_gc - 1) * gc_pitch + cell_gc.bbox().height() * dbu / 2)

    # --------------------------------------------------------
    # Place GC array
    # --------------------------------------------------------
    inst_GCs = coupler_array(
        cell,
        cell_name=gc_cell_name,
        cell_library=gc_library,
        x_offset=x_pos_GC / dbu,
        y_offset=y_pos_GC / dbu,
        label=device_label,
        label_location=label_location,
        count=num_gc,
    )

    if num_gc < 2:
        raise ValueError("num_gc must be at least 2.")

    # --------------------------------------------------------
    # Create / load device cell
    # If device_params is given, treat as PCell
    # Otherwise load as normal library cell
    # --------------------------------------------------------
    if device_params:
        cell_device = ly.create_cell(
            device_cell_name,
            device_library,
            device_params
        )
    else:
        cell_device = create_cell2(ly, device_cell_name, device_library)

    if cell_device is None:
        raise ValueError(
            f"Could not create/load device cell: "
            f"name='{device_cell_name}', library='{device_library}', params={device_params}"
        )

    # --------------------------------------------------------
    # Default device position
    # --------------------------------------------------------
    if device_x is None:
        device_x = inst_GCs[1].bbox().right * dbu + 4 * wg_bend_radius

    if device_y is None:
        device_y = inst_GCs[1].bbox().center().y * dbu#y_pos_GC - 0.5 * gc_pitch

    gc1 = inst_GCs[1]

    # print("left   =", gc1.bbox().left * dbu)
    # print("right  =", gc1.bbox().right * dbu)
    # print("bottom =", gc1.bbox().bottom * dbu)
    # print("top    =", gc1.bbox().top * dbu)

    # print("center x =", gc1.bbox().center().x * dbu)
    # print("center y =", gc1.bbox().center().y * dbu)

    t_dev = Trans(
        device_rotation,
        device_x / dbu,
        device_y / dbu
    )
    # inst_device = cell.insert(CellInstArray(cell_device.cell_index(), t_dev))

    inst_device_up = connect_cell(inst_GCs[0], "opt1", cell_device, "opt1",translation=device_rotation)
    inst_device_up.transform(Trans(50/dbu,0))
    
    inst_device_dw = connect_cell(inst_GCs[1], "opt1", cell_device, "opt1",translation=device_rotation)
    inst_device_dw.transform(Trans(50/dbu,0))

    
    # --------------------------------------------------------
    # Route GC[1] to device input
    # --------------------------------------------------------
    # connect_pins_with_waveguide(
    #     inst_GCs[3], gc_ports[3],
    #     inst_GCs[2], gc_ports[2],
    #     # turtle_A=routing_to_device,
    #     waveguide_type=wg_type,
    #     # r=wg_bend_radius
    # )
    
    connect_pins_with_waveguide(
        inst_GCs[1], gc_ports[1],
        inst_device_dw, "opt1",
        # inst_GCs[1], gc_ports[1],
        # turtle_A=routing_to_device,
        waveguide_type=wg_type,
        # r=wg_bend_radius
    )

    # --------------------------------------------------------
    # Route device output to GC[-1]
    # --------------------------------------------------------
    connect_pins_with_waveguide(
        # inst_device, "opt2",
        inst_GCs[0], gc_ports[0],
        inst_device_up, "opt1",

        # turtle_A=routing_from_device,
        waveguide_type=wg_type,
        r=wg_bend_radius
    )



    # --------------------------------------------------------
    # Suggested next placement position
    # --------------------------------------------------------
    next_x = cell.bbox().right * dbu + x_spacing + cell_gc.bbox().width() * dbu
    next_y = y_pos_GC

    if next_x + cell.bbox().width() * dbu > xmax:
        next_x = cell_gc.bbox().width() * dbu
        next_y = cell.bbox().bottom * dbu - row_spacing - cell_gc.bbox().height() * dbu / 2
        # next_y = cell.bbox().top * dbu - 900
    return next_x, next_y, xmax
```

```python
def ArbSplitter_cascaded(
    x_pos_GC=0,
    y_pos_GC=None,
    cellname="cellname",
    device_label="opt_in_TE_1550_device_JSong_Cascade",
    xmax=None,
    num_gc=2,
    num_dev=4,                  # total number of devices, must be multiple of 4
    device_cell_name="y_d8_split9901",
    device_library=lib_name,
    device_params=None,
    gc_ports=None,
    wg_type=None,
    x_spacing=20.0,
    row_spacing=None,
    first_dev_dx=50.0,          # um, shift first device from GC
    chain_dx=50.0,              # um, shift each next device from previous one
):
    """
    Two parallel cascaded splitter chains:
        GC0 -> upper chain
        GC1 -> lower chain

    Rules
    -----
    - num_dev must be a multiple of 4
    - total devices are split equally into upper/lower chains
    - adjacent devices are connected alternating:
        opt1-opt1, opt2-opt2, opt1-opt1, opt2-opt2, ...
    """

    global GC_SETTINGS, WG_SETTINGS

    if device_params is None:
        device_params = {}

    # -----------------------------
    # Validate inputs
    # -----------------------------
    if num_gc < 2:
        raise ValueError("num_gc must be at least 2.")

    if num_dev % 4 != 0:
        raise ValueError("num_dev must be a multiple of 4.")

    n_per_chain = num_dev // 2   # upper + lower chains

    # -----------------------------
    # Use global settings
    # -----------------------------
    gc_cell_name   = GC_SETTINGS["gc_cell_name"]
    gc_library     = GC_SETTINGS["gc_library"]
    gc_pitch       = GC_SETTINGS["gc_pitch"]
    label_location = GC_SETTINGS["label_location"]

    if gc_ports is None:
        gc_ports = ["opt1"] * num_gc
    elif len(gc_ports) != num_gc:
        raise ValueError("gc_ports length must equal num_gc")

    if wg_type is None:
        wg_type = WG_SETTINGS["wg_type"]

    if y_pos_GC is None:
        y_pos_GC = top_cell.bbox().height() * dbu

    if xmax is None:
        xmax = top_cell.bbox().width() * dbu

    if row_spacing is None:
        row_spacing = (num_gc - 1) * gc_pitch + 2 * wg_bend_radius

    # -----------------------------
    # Create new cell
    # -----------------------------
    cell = ly.create_cell(cellname)
    top_cell.insert(CellInstArray(cell.cell_index(), Trans(Trans.R0, 0, 0)))

    # -----------------------------
    # Load GC cell
    # -----------------------------
    cell_gc = create_cell2(ly, gc_cell_name, gc_library)
    if cell_gc is None:
        raise ValueError(f"Could not load GC cell '{gc_cell_name}' from library '{gc_library}'")

    # -----------------------------
    # Auto place first structure near top-left
    # -----------------------------
    if x_pos_GC == 0 and y_pos_GC == top_cell.bbox().height() * dbu:
        x_pos_GC = cell_gc.bbox().width() * dbu
        y_pos_GC -= (num_gc - 1) * gc_pitch + cell_gc.bbox().height() * dbu / 2

    # -----------------------------
    # Place GC array
    # -----------------------------
    inst_GCs = coupler_array(
        cell,
        cell_name=gc_cell_name,
        cell_library=gc_library,
        x_offset=x_pos_GC / dbu,
        y_offset=y_pos_GC / dbu,
        label=device_label,
        label_location=label_location,
        count=num_gc,
    )

    # -----------------------------
    # Load device cell
    # -----------------------------
    if device_params:
        cell_device = ly.create_cell(
            device_cell_name,
            device_library,
            device_params
        )
    else:
        cell_device = create_cell2(ly, device_cell_name, device_library)

    if cell_device is None:
        raise ValueError(
            f"Could not create/load device cell: "
            f"name='{device_cell_name}', library='{device_library}', params={device_params}"
        )

    # ---------------------------------------------------------
    # Helper: build one chain from one GC
    # ---------------------------------------------------------
    def build_chain(anchor_inst, anchor_port, name_prefix):
        """
        Build a cascaded chain starting from one GC instance.
        Returns list of device instances.
        """
        chain = []

        # ---- first device: anchor from GC opt1 to device opt1
        dev0 = connect_cell(
            anchor_inst,
            anchor_port,
            cell_device,
            "opt1",
            translation=Trans.R0
        )
        dev0.transform(Trans(int(first_dev_dx / dbu), 0))

        connect_pins_with_waveguide(
            anchor_inst, anchor_port,
            dev0, "opt1",
            waveguide_type=wg_type,
            r=wg_bend_radius
        )

        chain.append(dev0)

        # ---- remaining devices
        # print(f"n_per_chain is {n_per_chain}")

        for k in range(1, n_per_chain):
            k = k
            # print(f"k is {k}")

            prev_dev = chain[-1]

            # alternate same-port connections
            # k = 1 -> opt1-opt1
            # k = 2 -> opt2-opt2
            # k = 3 -> opt1-opt1
            # ...
            same_port = "opt1" if (k % 2 != 1) else "opt3"
            # print(f"{same_port}")
            if same_port == "opt1":
                rot = Trans.R0
                new_dev = connect_cell(
                    prev_dev,
                    same_port,
                    cell_device,
                    same_port,
                    translation=rot
                )
            else:
                rot = Trans.M90 #if (k%2  == 0) else Trans.R0
                x_prev = prev_dev.bbox().right * dbu-0.05+prev_dev.bbox().width() * dbu*1-0.1-1
                # print(f"right {prev_dev.bbox().right * dbu}")
                # print(f"width {prev_dev.bbox().width() * dbu}")
                # print(f"x_prev {x_prev}")
             
                y_prev = prev_dev.bbox().center().y * dbu

                t = Trans(rot, int(x_prev/dbu), int(y_prev/dbu))

                new_dev = cell.insert(CellInstArray(cell_device.cell_index(), t))
                 
            # for opt1-opt1, flip the new device
            # for opt2-opt2, keep default
            # rot = Trans.R0 if same_port == "opt1" else Trans.M0

            # new_dev = connect_cell(
            #     prev_dev,
            #     same_port,
            #     cell_device,
            #     same_port,
            #     translation=rot
            # )
            
        
            # new_dev.transform(Trans(int(chain_dx / dbu), 0))

            # connect_pins_with_waveguide(
            #     prev_dev, same_port,
            #     new_dev, same_port,
            #     waveguide_type=wg_type,
            #     r=wg_bend_radius
            # )

            chain.append(new_dev)

        return chain

    # -----------------------------
    # Build upper and lower chains
    # -----------------------------
    upper_chain = build_chain(inst_GCs[0], gc_ports[0], "UP")
    lower_chain = build_chain(inst_GCs[1], gc_ports[1], "DW")

    connect_pins_with_waveguide(
            upper_chain[-1], "opt1",
            lower_chain[-1], "opt1",
            waveguide_type=wg_type,
            r=wg_bend_radius
        )

    # -----------------------------
    # Suggested next placement
    # -----------------------------
    next_x = cell.bbox().right * dbu + x_spacing + cell_gc.bbox().width() * dbu
    next_y = y_pos_GC

    if next_x + cell.bbox().width() * dbu > xmax:
        next_x = cell_gc.bbox().width() * dbu
        next_y = cell.bbox().bottom * dbu - row_spacing - cell_gc.bbox().height() * dbu / 2

    return next_x, next_y, xmax, upper_chain, lower_chain
```

```python
import numpy as np
from pya import Trans, CellInstArray

# ============================================================
# Generic function to create one GC + splitter + device layout
# ============================================================
def place_device_with_gc_array(
    x_pos_GC=0,
    y_pos_GC=None,
    cellname="device_cell",
    device_label="opt_in_TE_1550_device",
    xmax=None,
    num_gc=4,
    gc_cell_name="ebeam_GC_Air_te1550",
    gc_library=None,
    ybranch_cell_name="ebeam_ybranch_te1550",
    ybranch_library="SiEPICfab_EBeam_ZEP_Beta",
    device_cell_name="ebeam_pcell_bragg_grating",
    device_library="SiEPICfab_EBeam_ZEP_Beta",
    device_params=None,
    device_rotation=Trans.R270,
    device_offset_x=None,
    device_offset_y=None,
    wg_type="Strip TE 1550 nm, w=500 nm (core-clad)",
    wg_width=0.5,
    GC_pitch=127.0,
    row_spacing=None,
    x_spacing=20.0,
):
    """
    Generic placement function for:
        GC array -> ybranch -> device -> ybranch -> routing

    Parameters
    ----------
    x_pos_GC, y_pos_GC : float
        Starting position of the grating coupler array in microns.
        If x_pos_GC == 0 and y_pos_GC is top boundary, auto-place near top-left.

    cellname : str
        Name of the new device cell.

    device_label : str
        Label placed with the GC array.

    xmax : float
        Maximum allowed x extent in microns before wrapping to next row.

    num_gc : int
        Number of grating couplers in the array.

    gc_cell_name, gc_library : str
        GC cell information.

    ybranch_cell_name, ybranch_library : str
        Y-branch cell information.

    device_cell_name, device_library : str
        Device cell information.

    device_params : dict
        PCell parameters for the device.
        Example:
            {
                "number_of_periods": 1000,
                "grating_period": 0.32,
                "corrugation_width": 0.02
            }

    device_rotation : Trans rotation
        Rotation of the device instance, e.g. Trans.R0, Trans.R90, Trans.R180, Trans.R270.

    device_offset_x, device_offset_y : float or None
        Manual offsets for device placement relative to the first ybranch.
        Units: microns.
        If None, default placement logic will be used.

    wg_type : str
        Waveguide type name in the PDK.

    GC_pitch : float
        Pitch between GCs in microns.

    row_spacing : float or None
        Vertical spacing between wrapped rows in microns.
        If None, a reasonable value is computed automatically.

    x_spacing : float
        Horizontal spacing between neighboring devices in microns.

    Returns
    -------
    next_x, next_y, xmax : float
        Suggested coordinates for placing the next device.
    """

    if gc_library is None:
        gc_library = tech_name

    if device_params is None:
        device_params = {}

    if y_pos_GC is None:
        y_pos_GC = top_cell.bbox().height() * dbu

    if xmax is None:
        xmax = top_cell.bbox().width() * dbu

    if row_spacing is None:
        row_spacing = (num_gc - 1) * GC_pitch + 2 * wg_bend_radius

    # --------------------------------------------------------
    # Create top-level child cell
    # --------------------------------------------------------
    cell = ly.create_cell(cellname)
    t = Trans(Trans.R0, 0, 0)
    top_cell.insert(CellInstArray(cell.cell_index(), t))

    # --------------------------------------------------------
    # Load PDK cells
    # --------------------------------------------------------
    cell_gc = create_cell2(ly, gc_cell_name, gc_library)
    cell_ybranch = create_cell2(ly, ybranch_cell_name, ybranch_library)

    # --------------------------------------------------------
    # Auto-position starting GC array near top-left
    # --------------------------------------------------------
    if x_pos_GC == 0 and y_pos_GC == top_cell.bbox().height() * dbu:
        x_pos_GC = cell_gc.bbox().width() * dbu
        y_pos_GC -= (num_gc - 1) * GC_pitch + cell_gc.bbox().height() * dbu / 2

    # --------------------------------------------------------
    # Place GC array
    # --------------------------------------------------------
    inst_GCs = coupler_array(
        cell,
        cell_name=gc_cell_name,
        cell_library=gc_library,
        x_offset=x_pos_GC / dbu,
        y_offset=y_pos_GC / dbu,
        label=device_label,
        label_location=2,
        count=num_gc,
    )

    # Safety check
    if num_gc < 4:
        raise ValueError("num_gc must be at least 4 for the current routing scheme.")

    # --------------------------------------------------------
    # First y-branch connected to GC[1]
    # --------------------------------------------------------
    inst_ybranch = connect_cell(inst_GCs[1], "opt1", cell_ybranch, "opt1")

    # --------------------------------------------------------
    # Create generic device cell
    # --------------------------------------------------------
    cell_device = ly.create_cell(
        device_cell_name,
        device_library,
        device_params
    )

    # --------------------------------------------------------
    # Place device relative to first y-branch
    # --------------------------------------------------------
    if device_offset_x is None:
        device_offset_x = (
            inst_ybranch.bbox().right * dbu
            + wg_bend_radius
            + inst_ybranch.bbox().width() * dbu / 2
        )

    if device_offset_y is None:
        device_offset_y = (
            inst_ybranch.bbox().bottom * dbu
            - inst_ybranch.bbox().height() * dbu
            - wg_bend_radius
        )

    t_dev = Trans(
        device_rotation,
        int(device_offset_x / dbu),
        int(device_offset_y / dbu)
    )
    inst_device = cell.insert(CellInstArray(cell_device.cell_index(), t_dev))

    # --------------------------------------------------------
    # Second y-branch connected to device
    # --------------------------------------------------------
    inst_ybranch2 = connect_cell(inst_device, "opt1", cell_ybranch, "opt1")

    # --------------------------------------------------------
    # Interconnect components
    # --------------------------------------------------------
    connect_pins_with_waveguide(
        inst_GCs[0], "opt1",
        inst_ybranch2, "opt3",
        waveguide_type=wg_type,
        r=wg_bend_radius
    )

    connect_pins_with_waveguide(
        inst_ybranch, "opt2",
        inst_ybranch2, "opt2",
        waveguide_type=wg_type,
        r=wg_bend_radius
    )

    connect_pins_with_waveguide(
        inst_GCs[2], "opt1",
        inst_ybranch, "opt3",
        waveguide_type=wg_type,
        r=wg_bend_radius
    )

    connect_pins_with_waveguide(
        inst_GCs[3], "opt1",
        inst_device, "opt2",
        turtle_A=[
            wg_bend_radius, -90,
            wg_bend_radius,  90
        ],
        waveguide_type=wg_type,
        r=wg_bend_radius
    )

    # --------------------------------------------------------
    # Placement for next device
    # --------------------------------------------------------
    next_x = cell.bbox().right * dbu + x_spacing + cell_gc.bbox().width() * dbu
    next_y = y_pos_GC

    # Wrap to next row if exceeding floorplan width
    if next_x + cell.bbox().width() * dbu > xmax:
        next_x = cell_gc.bbox().width() * dbu
        next_y = cell.bbox().bottom * dbu - row_spacing - cell_gc.bbox().height() * dbu / 2

    return next_x, next_y, xmax
```

```python
# Start near top-left of floorplan
x = 0
y = top_cell.bbox().height() * dbu
xmax = top_cell.bbox().width() * dbu

# Define grating period range (in microns)
grating_periods = np.arange(0.310, 0.340, 1e-3)

for gp in grating_periods:
    for k in range(1, 4):
        gp_nm = int(gp * 1000)

        cellname = f"ebeam_bragg_apodized_{gp_nm}_{k}"
        label = f"opt_in_TE_1550_device_Mateo_bragg_1000N_{gp_nm}nmPeriod_500nmW_20nmdW_{k}"

        x, y, xmax = place_device_with_gc_array(
            x_pos_GC=x,
            y_pos_GC=y,
            cellname=cellname,
            device_label=label,
            xmax=xmax,
            num_gc=4,
            gc_cell_name="ebeam_GC_Air_te1550",
            gc_library=tech_name,
            ybranch_cell_name="ebeam_ybranch_te1550",
            ybranch_library="SiEPICfab_EBeam_ZEP_Beta",
            device_cell_name="ebeam_pcell_bragg_grating",
            device_library="SiEPICfab_EBeam_ZEP_Beta",
            device_params={
                "number_of_periods": 1000,
                "grating_period": gp,
                "corrugation_width": 0.02
            },
            device_rotation=Trans.R270
        )
```
